```python
import glob
import os
import numpy as np
from sklearn.preprocessing import StandardScaler
import joblib

# Get a list of task folders (e.g., /data/hbz15/HIPPO-SVGP_data/data/era5/processed_timeseries/task_1, ..., task_10)
task_folders = [
    # "processed_timeseries/task_1",
    # "processed_timeseries/task_2",
    # "processed_timeseries/task_3",
    # "processed_timeseries/task_4",
    # "processed_timeseries/task_5",
    # "processed_timeseries/task_6",
    # "processed_timeseries/task_7",
    # "processed_timeseries/task_8",
    # "processed_timeseries/task_9",
    # "processed_timeseries/task_10",
    # "processed_timeseries_2/task_1",
    # "processed_timeseries_2/task_2",
    # "processed_timeseries_2/task_3",
    # "processed_timeseries_2/task_4",
    # "processed_timeseries_2/task_5",
    # "processed_timeseries_2/task_6",
    # "processed_timeseries_2/task_7",
    # "processed_timeseries_2/task_8",
    # "processed_timeseries_2/task_9",
    # "processed_timeseries_2/task_10",
    # "processed_timeseries_2/task_11",
    # "processed_timeseries_2/task_12",
    # "processed_timeseries_2/task_13",
    # "processed_timeseries_2/task_14",
    # "processed_timeseries_2/task_15",
    # "processed_timeseries_2/task_16",
    # "processed_timeseries_2/task_17",
    # "processed_timeseries_2/task_18",
    # "processed_timeseries_2/task_19",
    # "processed_timeseries_2/task_20",
    # "processed_timeseries_2/task_21",
    # "processed_timeseries_2/task_22",
    # "processed_timeseries_2/task_23",
    # "processed_timeseries_2/task_24",
    # "processed_timeseries_2/task_25",
    "processed_timeseries_4/task_1",
    "processed_timeseries_4/task_2",
    "processed_timeseries_4/task_3",
    "processed_timeseries_4/task_4",
    "processed_timeseries_4/task_5",
    "processed_timeseries_4/task_6",
    "processed_timeseries_4/task_7",
    "processed_timeseries_4/task_8",
    "processed_timeseries_4/task_9",
    "processed_timeseries_4/task_10",
    "processed_timeseries_4/task_11",
    "processed_timeseries_4/task_12",
    "processed_timeseries_4/task_13",
    "processed_timeseries_4/task_14",
    "processed_timeseries_4/task_15",
    "processed_timeseries_4/task_16",
    "processed_timeseries_4/task_17",
    "processed_timeseries_4/task_18",
    "processed_timeseries_4/task_19",
    "processed_timeseries_4/task_20",
    "processed_timeseries_4/task_21",
    "processed_timeseries_4/task_22",
    "processed_timeseries_4/task_23",
    "processed_timeseries_4/task_24",
    "processed_timeseries_4/task_25",
]

# Extract base directory path
base_dir = os.path.dirname(task_folders[0])

# First, process task_1 to create the scaler that will be used for all tasks
task1_folder = task_folders[0]
print(f"\nProcessing task 1 folder to create global scaler: {task1_folder}")

# Get all .npz files in task 1 folder
task1_files = glob.glob(os.path.join(task1_folder, "sequences", "*.npz"))

# Concatenate all training data from files in task 1
train_data_all = []
for task_file in task1_files:
    npz_data = np.load(task_file)
    # Assume 'data_train' has shape (features, n_train_samples)
    train_data = npz_data['data_train']
    # Transpose to have samples as rows for the scaler
    train_data_all.append(train_data.T)

if not train_data_all:
    print(f"No training data found in {task1_folder}. Cannot proceed without task 1 data.")
else:
    # Concatenate along axis 0 (stacking samples from all files)
    concatenated_train = np.concatenate(train_data_all, axis=0)
    print("Task 1 concatenated training data shape:", concatenated_train.shape)
    
    # Fit the scaler on the concatenated training data from task 1
    global_scaler = StandardScaler()
    global_scaler.fit(concatenated_train)
    
    # Save the fitted scaler to task 1 folder and to the base directory (for easy access)
    task1_scaler_file = os.path.join(task1_folder, "scaler.pkl")
    global_scaler_file = os.path.join(base_dir, "global_scaler.pkl")
    joblib.dump(global_scaler, task1_scaler_file)
    joblib.dump(global_scaler, global_scaler_file)
    print(f"Saved global scaler to {task1_scaler_file} and {global_scaler_file}")
    
    # Now loop through all task folders and use the scaler from task 1
    for task_folder in task_folders:
        print(f"\nProcessing folder: {task_folder}")
        
        # Get all .npz files in the current task folder
        task_files = glob.glob(os.path.join(task_folder, "sequences", "*.npz"))
        
        # Create a reference to the scaler (or save a copy) in the current task folder
        task_scaler_file = os.path.join(task_folder, "scaler.pkl")
        if task_folder != task1_folder:  # Don't overwrite task 1's scaler
            joblib.dump(global_scaler, task_scaler_file)
            print(f"Saved reference to global scaler in {task_scaler_file}")
        
        # Apply the global scaler to each file in the current task folder
        for task_file in task_files:
            npz_data = np.load(task_file)
            data_train = npz_data['data_train']   # shape: (features, n_train_samples)
            data_val   = npz_data['data_val']     # shape: (features, n_val_samples)
            data_test  = npz_data['data_test']    # shape: (features, n_test_samples)
            
            # Scale each split using the global scaler
            scaled_train = global_scaler.transform(data_train.T).T
            scaled_val   = global_scaler.transform(data_val.T).T
            scaled_test  = global_scaler.transform(data_test.T).T
            
            # Create a new filename for the scaled data (e.g., appending '_scaled' to the filename)
            scaled_file = task_file.replace(".npz", "_scaled.npz")
            np.savez(scaled_file, 
                     data_train=scaled_train, time_train=npz_data['time_train'],
                     data_val=scaled_val, time_val=npz_data['time_val'],
                     data_test=scaled_test, time_test=npz_data['time_test'])
            
            print(f"Saved scaled data to {scaled_file}")
```
